```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
classifier([
    "This course is so much fun to go through!",
    "This course is extremely boring"
])
```

```python
zero_shot_classifier = pipeline("zero-shot-classification")
zero_shot_classifier(
    "I am learning how to use the Hugging Face transformers library!",
    candidate_labels = ["Machine Learning", "Mathematics", "Literature"]
)
```

```python
mask = pipeline("fill-mask", model="bert-base-cased")
mask("This course will teach you all about [MASK] models.", top_k=2)
```
