exploration vs. exploitation, use things you already know to your advantage or explore (which is riskier).

policy is the function that determines what action to take at a certain state. Can learn which action to take **(policy based method)**, or which state is more valuable to the agent tries to get to that state **(value based method)**

**deterministic** means that the policy will output the same action every time given the same state
a = pi(s)

**stochastic** outputs a probability distribution over the actions.
pi(a|s) = P[A|s]

In value based methods, instead of learning the policy function, you learn a value function that maps a state to its expected value
The value of a state is the expected discounted return the agent can get if it starts in that state, and then acts according to our policy.
